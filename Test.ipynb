```python
from datetime import datetime
from pathlib import Path
import os
import subprocess
import sys

PROJECT_ROOT = Path(
    r"D:\Anaconda Python Files\GitHub\PINN-Thesis-Framework-MPhil-2026-2027-"
)
TABLES = PROJECT_ROOT / "results" / "tables"

inputs = {
    "--ghana": [
        TABLES / "grid_ghana_split.csv",
        TABLES / "grid_ghana_joint.csv",
    ],
    "--italy-primary": [
        TABLES / "italy_primary_grid_split.csv",
        TABLES / "italy_primary_grid_joint.csv",
    ],
    "--italy-secondary": [
        TABLES / "italy_secondary_grid_split.csv",
        TABLES / "italy_secondary_grid_joint.csv",
    ],
}

script = PROJECT_ROOT / "src" / "fig_degradation_equity.py"
required = [script] + [
    path for paths in inputs.values() for path in paths
]
missing = [str(path) for path in required if not path.is_file()]

if missing:
    raise FileNotFoundError("Missing files:\n" + "\n".join(missing))

# Ensure this notebook is using the dedicated plotting environment.
print("Python:", sys.executable)
if Path(sys.prefix).name.lower() != "pinn-plots":
    raise RuntimeError("Select the pinn-plots kernel before running this cell")

output_dir = (
    PROJECT_ROOT / "results" / "figures" 
    / datetime.now().strftime("isolated_%Y%m%d_%H%M%S_%f")
)

command = [sys.executable, "-u", str(script)]
for flag, paths in inputs.items():
    command.extend([flag, *map(str, paths)])
command.extend(["--outdir", str(output_dir)])

# Enable only after verifying the legacy labels in each input file.
ALLOW_LEGACY_ITALY_LABEL = True
if ALLOW_LEGACY_ITALY_LABEL:
    command.append("--allow-legacy-italy-label")

env = os.environ.copy()
env["MPLBACKEND"] = "Agg"

print("Running:", subprocess.list2cmdline(command), flush=True)

result = subprocess.run(
    command,
    cwd=PROJECT_ROOT,
    env=env,
    capture_output=True,
    text=True,
)

print(result.stdout)
if result.stderr:
    print(result.stderr)

result.check_returncode()
print("Saved outputs:", output_dir)
```
